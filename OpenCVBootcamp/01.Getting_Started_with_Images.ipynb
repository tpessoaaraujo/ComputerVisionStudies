```python
# Lendo a imagem
coke_img = cv2.imread("assets/coca-cola-logo.png")

# Exibindo o tamanho da imagem
print("O tamanho da imagem é: ", coke_img.shape)

# Exibindo o tipo de dados da imagem
print("O tipo de dados da imagem é: ", coke_img.dtype)
```

```python
# Echibindo a imagem usando Matplotlib
plt.imshow(coke_img)
```

```python
# A cor da imagem plotada usando o Matplotlib é diferente da imagem original. Isso ocorre porque o OpenCV lê as imagens no formato BGR (Blue, Green, Red), enquanto o Matplotlib exibe as imagens no formato RGB (Red, Green, Blue). Para corrigir isso, precisamos converter a imagem do formato BGR para RGB antes de exibi-la usando o Matplotlib.
```

```python
# Convertendo a imagem do formato BGR para RGB
coke_img_channels_reversed = coke_img[:, :, ::-1]

# Exibindo a imagem convertida usando Matplotlib
plt.imshow(coke_img_channels_reversed)
```

```python
# Splitando os canais de cor da imagem
img_NZ_bgr = cv2.imread("assets/New_Zealand_Lake.jpg", cv2.IMREAD_COLOR)
b,g,r = cv2.split(img_NZ_bgr)

# Exibindo os canais de cor
plt.figure(figsize=[20, 5])
plt.subplot(141);plt.imshow(r);plt.title("Canal Vermelho")
plt.subplot(142);plt.imshow(g);plt.title("Canal Verde")
plt.subplot(143);plt.imshow(b);plt.title("Canal Azul")

# Juntando os canais de cor para formar a imagem original
imgMerged = cv2.merge((b,g,r))

# Exibindo a imagem resultante da junção dos canais
plt.subplot(144);plt.imshow(imgMerged);plt.title("Saída da junção dos canais")
```

```python
# Exibindo os canais de cor em escala de cinza
plt.figure(figsize=[20, 5])
plt.subplot(141);plt.imshow(r, cmap="gray");plt.title("Canal Vermelho em escala de cinza")
plt.subplot(142);plt.imshow(g, cmap="gray");plt.title("Canal Verde em escala de cinza")
plt.subplot(143);plt.imshow(b, cmap="gray");plt.title("Canal Azul em escala de cinza")

# Juntando os canais de cor para formar a imagem original
imgMerged = cv2.merge((b,g,r))

# Exibindo a imagem resultante da junção dos canais em RGB
plt.subplot(144);plt.imshow(imgMerged[:,:,::-1]);plt.title("Saída da junção dos canais em RGB")
```

```python
# Convertendo a imagem do formato BGR para RGB
img_NZ_rgb = cv2.cvtColor(img_NZ_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_NZ_rgb)
```

```python
img_hsv = cv2.cvtColor(img_NZ_bgr, cv2.COLOR_BGR2HSV)
h,s,v = cv2.split(img_hsv)

# Exibindo os canais HSV em escala de cinza
plt.figure(figsize=[20, 5])
plt.subplot(141);plt.imshow(h, cmap="gray");plt.title("Canal H (Matiz)")
plt.subplot(142);plt.imshow(s, cmap="gray");plt.title("Canal S (Saturação)")
plt.subplot(143);plt.imshow(v, cmap="gray");plt.title("Canal V (Valor)")
plt.subplot(144);plt.imshow(img_NZ_rgb);plt.title("Imagem RGB Original")
```

```python
# Modificando o canal individualmente
h_new = h + 10
img_NZ_merged = cv2.merge((h_new, s, v))
img_NZ_rgb = cv2.cvtColor(img_NZ_merged, cv2.COLOR_HSV2RGB)

# Exibindo os canais modificados
plt.figure(figsize=[20, 5])
plt.subplot(141);plt.imshow(h_new, cmap="gray");plt.title("Canal H modificado +10 (Matiz)")
plt.subplot(142);plt.imshow(s, cmap="gray");plt.title("Canal S (Saturação)")
plt.subplot(143);plt.imshow(v, cmap="gray");plt.title("Canal V (Valor)")
plt.subplot(144);plt.imshow(img_NZ_rgb);plt.title("Imagem RGB modificada")
```

```python
# Salvando a imagem
cv2.imwrite("outputs/New_Zealand_Lake_SAVED.png", img_NZ_bgr)
Image(filename='outputs/New_Zealand_Lake_SAVED.png')
```
