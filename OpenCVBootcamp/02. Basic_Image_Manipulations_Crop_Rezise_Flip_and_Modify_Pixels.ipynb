```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import Image
%matplotlib inline
```

```python
# Lendo a imagem em escala de cinza
cb_img = cv2.imread("assets/checkerboard_18x18.png", 0)

# Definindo o mapa de cores para tons de cinza para uma renderização adequada.
plt.imshow(cb_img, cmap="gray")
print(cb_img)
```

```python
# Imprimindo o primeiro pixel do primeiro quadrado preto
print(cb_img[0, 0])

# Imprimindo o primeiro pixel branco a direita do primeiro quadrado preto
print(cb_img[0, 6])
```

```python
# Modificando os pixels da imagem
cb_img_copy = cb_img.copy()
cb_img_copy[2, 2] = 200
cb_img_copy[2, 3] = 200
cb_img_copy[3, 2] = 200
cb_img_copy[3, 3] = 200

plt.imshow(cb_img_copy, cmap="gray")
print(cb_img_copy)
```

```python
img_NZ_bgr = cv2.imread("assets/New_Zealand_Boat.jpg", cv2.IMREAD_COLOR)
img_NZ_rgb = img_NZ_bgr[:, :, ::-1]
plt.imshow(img_NZ_rgb)
```

```python
# Cortando Imagens
cropeed_region = img_NZ_rgb[200:400, 300:600]
plt.imshow(cropeed_region)
```

```python
# Redimensando Imagens
# Método 1: Redimensionar para um tamanho específico usando fx e fy
resized_cropeed_region_2x = cv2.resize(cropeed_region, None, fx=2, fy = 2)
plt.imshow(resized_cropeed_region_2x)
```

```python
# Método 2: Redimensionar para um tamanho específico usando width e height
desired_width = 100
desired_height = 200
dim = (desired_width, desired_height)

resized_cropeed_region = cv2.resize(cropeed_region, dsize=dim, interpolation=cv2.INTER_AREA)
plt.imshow(resized_cropeed_region)
```

```python
# Método 3: Redimensionar mantendo a proporção
desired_width = 100
aspect_ratio = desired_width / cropeed_region.shape[1]
desired_height = int(cropeed_region.shape[0] * aspect_ratio)
dim = (desired_width, desired_height)

resized_cropeed_region = cv2.resize(cropeed_region, dsize=dim, interpolation=cv2.INTER_AREA)
plt.imshow(resized_cropeed_region)
```

```python
# Alterando a ordem dos canais
resized_cropeed_region_2x = resized_cropeed_region_2x[:, :, ::-1]

# Salvando a imagem redimensionada
cv2.imwrite("assets/resized_cropeed_region_2x.png", resized_cropeed_region_2x)

# Mostrando a imagem cortada e redimensionada
Image(filename="assets/resized_cropeed_region_2x.png")
```

```python
# Rotacionando a imagem
img_NZ_rgb_flipped_horz = cv2.flip(img_NZ_rgb, 1)
img_NZ_rgb_flipped_vert = cv2.flip(img_NZ_rgb, 0)
img_NZ_rgb_flipped_both = cv2.flip(img_NZ_rgb, -1)

# Mostrando as imagens rotacionadas
plt.figure(figsize=(18, 5))
plt.subplot(141);plt.imshow(img_NZ_rgb_flipped_horz);plt.title("Flip Horizontal")
plt.subplot(142);plt.imshow(img_NZ_rgb_flipped_vert);plt.title("Flip Vertical")
plt.subplot(143);plt.imshow(img_NZ_rgb_flipped_both);plt.title("Flip Both")
plt.subplot(144);plt.imshow(img_NZ_rgb);plt.title("Original")
```
